## Análise de Temperatura Vertical

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df = df_final
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
df = df.sort_values('Timestamp')

scripts_unicos = df['Script'].unique()

fig, axes = plt.subplots(len(scripts_unicos), 1, figsize=(12, 15), sharex=False)

if len(scripts_unicos) == 1:
    axes = [axes]

colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

for i, script in enumerate(scripts_unicos):
    ax = axes[i]
    
    # Filtrar dados
    df_script = df[df['Script'] == script]
    
    # Plot Original
    ax.plot(df_script['Timestamp'], df_script['Temp°C'], 
            marker='o', linestyle='-', markersize=2, alpha=0.6, 
            color=colors[i % len(colors)], label='Temperatura')
    
    # --- NOVO: Calcular e Plotar a Média ---
    media_temp = df_script['Temp°C'].mean()
    
    # axhline desenha uma linha horizontal em todo o eixo
    ax.axhline(y=media_temp, color='red', linestyle='--', linewidth=2, 
               label=f'Média: {media_temp:.1f}°C')
   
    # Estilização
    ax.set_title(f'Evolução da Temperatura - Script: {script}', fontsize=12, fontweight='bold')
    ax.set_ylabel('Temperatura (°C)')
    ax.legend() # Importante adicionar a legenda para mostrar o valor da média
    ax.grid(True, linestyle='--', alpha=0.6)
    
    # Formatação de Data
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M:%S'))
    #plt.setp(ax.get_xticklabels(), rotation=45, ha='right')

plt.savefig('Relatorio/images/analise_temperatura.pdf', format='pdf', bbox_inches='tight')

plt.xlabel('Horário de Execução')
plt.tight_layout()
plt.show()

```

## Análise de Temperatura Compacta - Horizontal

```python
import scipy.stats as st

# Configuração para 1 linha e 3 colunas
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(15, 4), sharey=True)

if len(scripts_unicos) == 1:
    axes = [axes]

colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

for i, script in enumerate(scripts_unicos):
    ax = axes[i]
    df_script = df[df['Script'] == script]

    # Série de temperatura e timestamps
    temps = df_script['Temp°C']
    ts = df_script['Timestamp']

    # ---- Cálculo do intervalo de confiança 95% ----
    mean_temp = temps.mean()
    std_temp = temps.std(ddof=1)
    n = len(temps)

    sem_temp = std_temp / (n ** 0.5)  # erro padrão
    ci_low, ci_high = st.t.interval(
        0.95,
        df=n - 1,
        loc=mean_temp,
        scale=sem_temp
    )

    # Banda de intervalo de confiança (linha horizontal grossa)
    ax.fill_between(
        ts,
        ci_low,
        ci_high,
        color=colors[i % len(colors)],
        alpha=0.15,
        label=f'IC 95%: [{ci_low:.1f}, {ci_high:.1f}]'
    )

    # ---- Plot da temperatura medida ----
    ax.plot(ts, temps,
            marker='o', linestyle='-', markersize=2, alpha=0.6,
            color=colors[i % len(colors)])

    # Linha da média
    ax.axhline(y=mean_temp, color='red', linestyle='--', linewidth=1.5,
               label=f'Média: {mean_temp:.1f}°C')

    ax.set_title(script, fontsize=12, fontweight='bold')

    if i == 0:
        ax.set_ylabel('Temperatura (°C)')

    if i == 0 or i == 1:
        ax.legend(fontsize='small', loc='lower right')
    else:
        ax.legend(fontsize='small', loc='upper right')
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.set_xlabel('Horário de Execução', fontsize=10)

    ax.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))
    plt.setp(ax.get_xticklabels(), rotation=45, ha='right', fontsize=9)

plt.tight_layout()
plt.savefig('Relatorio/images/analise_temperatura_compacta.pdf', format='pdf', bbox_inches='tight')
plt.show()
```

## Tempor de execução por script

```python
import matplotlib.pyplot as plt
import seaborn as sns

# --- CONFIGURAÇÃO ---
# Ordem dos gráficos (Scripts) - Serão as colunas
scripts_order = ['Multiprocessing', 'Dask', 'Threading']

# Ordem das linhas (Cenários) - Será o 'hue'
cpu_order = ['8P_HT_OFF', '4P_HT_ON', '4P_4E_HT_OFF', '8E']

# Valores do eixo X
valores_eixo_x = [1000, 2000, 4000, 6000, 8000]

# Cria figura com 1 linha e 3 colunas
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Se axes não for lista (caso fosse 1 gráfico só), garantimos que seja iterável
if len(scripts_order) == 1:
    axes = [axes]

for i, script_name in enumerate(scripts_order):
    ax = axes[i]
    
    # Filtra o DF para pegar apenas dados desse Script específico
    df_subset = df[df['Script'] == script_name]
    
    # --- PLOTAGEM ---
    sns.lineplot(
        data=df_subset,
        x='Matrix_Size',
        y='Time_sec',
        hue='CPU_Set_Name',      # AQUI MUDOU: As cores agora são os Cenários
        style='CPU_Set_Name',    # Estilos de linha diferentes para cada cenário
        hue_order=cpu_order,     # Garante a ordem das cores
        style_order=cpu_order,   # Garante a ordem dos estilos
        markers=True,
        dashes=False,
        ax=ax,
        errorbar=('ci', 95)      # Intervalo de confiança
    )
    
    # --- FORMATAÇÃO ---
    ax.set_xticks(valores_eixo_x)
    ax.set_title(f'Script: {script_name}')
    
    ax.set_xlabel('Tamanho da Matriz')
    
    # Apenas o primeiro gráfico ganha label Y para ficar limpo
    if i == 0:
        ax.set_ylabel('Tempo (s)')
    else:
        ax.set_ylabel(None)
        
    ax.grid(True, linestyle='--', alpha=0.3)
    
    # --- LEGENDA ---
    # Mostra a legenda apenas no primeiro gráfico (ou no último, como preferir)
    if i == 0:
        ax.legend(title='Cenário CPU', loc='upper left', fontsize='small')
    else:
        if ax.get_legend():
            ax.get_legend().remove()

plt.tight_layout()
plt.savefig('Relatorio/images/line_tempo_por_scripts.pdf', format='pdf', bbox_inches='tight')
plt.show()
```

## Tempo de execução por tamanho de matriz

### Bar Plot
- Barra colorida: média total
- Linha preta: Intervalo de confiança - intervalo dentro do qual a verdadeira média provavelmente está
    - Linha longa: provavelmente houve alguns outliers (execuções muito rápidas ou muito lentas)

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração da grade: 2 linhas x 3 colunas
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes_flat = axes.flatten()

# Definições
ordem_eixo_x = ['8P_HT_OFF', '4P_HT_ON', '4P_4E_HT_OFF', '8E']
matrix_sizes = [1000, 2000, 4000, 6000, 8000]

# Variáveis para guardar as referências da legenda
handles, labels = None, None

for i, size in enumerate(matrix_sizes):
    ax = axes_flat[i]
    
    # Filtra dados
    df_subset = df[df['Matrix_Size'] == size]
    
    # Lógica da Legenda:
    # Só geramos a legenda no primeiro gráfico (i==0) para capturar os dados.
    # Nos outros (else), já plotamos sem legenda para economizar processamento.
    if i == 0:
        sns.barplot(
            data=df_subset,
            x='CPU_Set_Name',
            y='Time_sec',
            hue='Script',
            order=ordem_eixo_x,
            capsize=0.1,
            errorbar=('ci', 95),
            ax=ax
        )
        # Captura as cores e nomes
        handles, labels = ax.get_legend_handles_labels()
        # Remove a legenda deste gráfico (queremos ela só no final)
        ax.get_legend().remove()
    else:
        sns.barplot(
            data=df_subset,
            x='CPU_Set_Name',
            y='Time_sec',
            hue='Script',
            order=ordem_eixo_x,
            capsize=0.1,
            errorbar=('ci', 95),
            legend=False, # Não gera legenda
            ax=ax
        )

    ax.set_title(f'Matriz: {size}')
    ax.set_ylabel('Tempo Médio (s)')
    ax.set_xlabel('Cenário de CPU')
    ax.grid(True, axis='y', linestyle='--', alpha=0.3)

# --- CONFIGURAÇÃO DO 6º ESPAÇO (A LEGENDA) ---

# Seleciona o eixo vazio (índice 5)
ax_legenda = axes_flat[5]

# Desliga as bordas e eixos (para ficar um fundo branco limpo)
ax_legenda.axis('off')

# Cria a legenda neste espaço usando os dados capturados do primeiro gráfico
if handles and labels:
    ax_legenda.legend(
        handles, 
        labels, 
        title='Script', 
        loc='center',   # Centraliza perfeitamente no espaço em branco
        fontsize='x-large', 
        title_fontsize='xx-large',
        frameon=False   # Remove a moldura da legenda (opcional, fica mais clean)
    )

#plt.suptitle('Tempo Médio de Execução com Intervalo de Confiança (95%)', fontsize=16)
plt.savefig('Relatorio/images/histograma_tempo_por_matix_size.pdf', format='pdf', bbox_inches='tight')

plt.tight_layout()
plt.show()
```
