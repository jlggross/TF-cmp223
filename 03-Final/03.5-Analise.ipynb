## Análise de Temperatura Vertical

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df = df_final
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
df = df.sort_values('Timestamp')

scripts_unicos = df['Script'].unique()

fig, axes = plt.subplots(len(scripts_unicos), 1, figsize=(12, 15), sharex=False)

if len(scripts_unicos) == 1:
    axes = [axes]

colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

for i, script in enumerate(scripts_unicos):
    ax = axes[i]
    
    # Filtrar dados
    df_script = df[df['Script'] == script]
    
    # Plot Original
    ax.plot(df_script['Timestamp'], df_script['Temp°C'], 
            marker='o', linestyle='-', markersize=2, alpha=0.6, 
            color=colors[i % len(colors)], label='Temperatura')
    
    # --- NOVO: Calcular e Plotar a Média ---
    media_temp = df_script['Temp°C'].mean()
    
    # axhline desenha uma linha horizontal em todo o eixo
    ax.axhline(y=media_temp, color='red', linestyle='--', linewidth=2, 
               label=f'Média: {media_temp:.1f}°C')
   
    # Estilização
    ax.set_title(f'Evolução da Temperatura - Script: {script}', fontsize=12, fontweight='bold')
    ax.set_ylabel('Temperatura (°C)')
    ax.legend() # Importante adicionar a legenda para mostrar o valor da média
    ax.grid(True, linestyle='--', alpha=0.6)
    
    # Formatação de Data
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M:%S'))
    #plt.setp(ax.get_xticklabels(), rotation=45, ha='right')

plt.savefig('Relatorio/images/analise_temperatura.pdf', format='pdf', bbox_inches='tight')

plt.xlabel('Horário de Execução')
plt.tight_layout()
plt.show()

```

## Análise de Temperatura Compacta - Horizontal

```python
# Configuração para 1 linha e 3 colunas
# figsize=(15, 4) cria uma figura larga e curta
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(15, 4), sharey=True) 

# Garante que axes seja iterável mesmo se houver apenas 1 script
if len(scripts_unicos) == 1:
    axes = [axes]

colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

for i, script in enumerate(scripts_unicos):
    ax = axes[i] # Acessa o eixo correspondente
    
    df_script = df[df['Script'] == script]
    
    ax.plot(df_script['Timestamp'], df_script['Temp°C'], 
            marker='o', linestyle='-', markersize=2, alpha=0.6, 
            color=colors[i % len(colors)])
    
    media_temp = df_script['Temp°C'].mean()
    ax.axhline(y=media_temp, color='red', linestyle='--', linewidth=1.5, 
               label=f'Média: {media_temp:.1f}°C')
    
    # Título mais curto para economizar espaço
    ax.set_title(script, fontsize=12, fontweight='bold')
    
    # Define label Y apenas no primeiro gráfico para não poluir
    if i == 0:
        ax.set_ylabel('Temperatura (°C)')
    
    ax.legend(fontsize='small', loc='upper right') # Legenda menor
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.set_xlabel('Horário de Execução', fontsize=10)
    
    # Formatação de hora simplificada
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))
    plt.setp(ax.get_xticklabels(), rotation=45, ha='right', fontsize=9)

# Ajuste fino das margens
plt.tight_layout()
plt.savefig('Relatorio/images/analise_temperatura_compacta.pdf', format='pdf', bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Supondo que seu DataFrame se chama 'df'
# df = pd.read_csv('seu_arquivo.csv')

sns.set_style("whitegrid")
unique_sizes = sorted(df['Matrix_Size'].unique())

# 1. Criar a grade padrão 2 linhas x 3 colunas
# Isso cria 6 espaços no total. Como temos 5 gráficos, sobrará 1.
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# 'axes' é uma matriz 2x3. Usamos .flatten() para transformar em uma lista simples [0, 1, 2, 3, 4, 5]
# facilitando o loop.
axes_flat = axes.flatten()

# 2. Loop de plotagem
for i, size in enumerate(unique_sizes):
    ax = axes_flat[i]
    subset = df[df['Matrix_Size'] == size]
    
    # Criar Boxplot
    sns.boxplot(x='Script', y='Time_sec', data=subset, ax=ax, hue='Script', palette="viridis")
    
    # Títulos e Rótulos
    ax.set_title(f'Matrix Size: {size}', fontsize=14, fontweight='bold')
    ax.set_xlabel('')
    
    # Label Y apenas na primeira coluna (índices 0 e 3 na lista flat)
    if i in [0, 3]:
        ax.set_ylabel('Tempo (segundos)', fontsize=12)
    else:
        ax.set_ylabel('')
        
    # Remove a legenda individual
    if ax.get_legend():
        ax.get_legend().remove()
    
    # Rotacionar labels do eixo X
    ax.tick_params(axis='x', rotation=45)

# 3. Remover o último gráfico vazio (o 6º espaço, índice 5)
fig.delaxes(axes_flat[5])

plt.tight_layout()
plt.show()
```
