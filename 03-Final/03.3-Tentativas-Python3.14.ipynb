# Python 3.14

- Necessário criar novo ipykernel para o jupyer Notebook
- Jupyter Nottebok visualiza por padrão apenas o ipykernel do sistema operacional, que é o Python 3.12


## pyenv e ipykernel
- Instalação do pyenv para definir Python 3.14 como ipykernel a ser utilziado pelo Jupyter Notebook
    - Python do sistema possui versão 3.12.3 e é utilziado pelo apt, interface gráfica e por ferramentas de sistema (/usr/bin/python3.12)
    - O Jupyer Notebook percebia apenas o Kernel Python como sendo o kernel Python 3.12
    - Ao instalar o Python 3.14.0, por estar a nível de usuário, não era refletido nokernel Python percebido pelo Jupyer
    - Para evitar quebrar o SO Ubuntu, não era recomendável simplesmente atualizar o python da versão 3.12 para versão 3.14. Foi necessário umas instalação paralela do 3.14
    - O pyenv é um gerenciador de contexto e ambiente e permitiu instalar e compilar código fonte do python 3.14, além de configurar juntamente ao .bashrc que o python global utilizado seria o 3.14

```bash
# Ajustes e limpeza
sudo timedatectl set-ntp on # Para evitar erros com apt
sudo apt update
sudo apt clean
sudo apt autoremove

# Dependências de compilação
sudo apt install -y build-essential libssl-dev zlib1g-dev \
libbz2-dev libreadline-dev libsqlite3-dev curl \
libncursesw5-dev xz-utils tk-dev libxml2-dev libxmlsec1-dev libffi-dev liblzma-dev

# Instalação do pyenv
git clone https://github.com/pyenv/pyenv.git ~/.pyenv

# Configuração do pyenv no Terminal
echo 'export PYENV_ROOT="$HOME/.pyenv"' >> ~/.bashrc
echo 'export PATH="$PYENV_ROOT/bin:$PATH"' >> ~/.bashrc
echo -e 'if command -v pyenv 1>/dev/null 2>&1; then\n  eval "$(pyenv init -)"\nfi' >> ~/.bashrc

# Usando o pyenv para instalar o python 3.14
pyenv install 3.14.0
pyenv global 3.14.0

# Instalação do ipykernel dentro da instalação do Python 3.14
pip --proxy "http://servidor.proxy:porta" install ipykernel

# Registro de novo kernel para o ambiente do Jupyer
python -m ipykernel install --user --name "python-3.14"
```

- Sem o pyenv, teríamos um Python 3.14 "órfão", e o Jupyter nunca o encontraria.
- O pyenv foi necessário porque ele é a única ferramenta que nos permite:
    1. Construir uma versão paralela do Python (3.14) sem quebrar o sistema (3.12).
    2. Ativar temporariamente esse Python 3.14 no terminal.
    3. Usar essa ativação para "apresentar" formalmente o 3.14 ao Jupyter, criando um novo Kernel que pudéssemos selecionar.
