```python
3 aplicações

MATRIX_SIZE = 1000, 2000, 4000, 6000

WAIT_TIME = 10 (segundos)

CPU_SETS = {   
    "8P_HT_OFF": [0, 2, 4, 6, 8, 10, 12, 14], # p-cores (8 cores, 1 thread per core)
    "4P_HT_ON": [0, 1, 2, 3, 4, 5, 6, 7], # p-cores (4 cores, 2 threads per core)
    "4P_4E_HT_OFF": [0, 2, 4, 6, 16, 17, 18, 19], # p-cores e e-cores (4 cores p-core, 4 cores e-core)
    "8E": [16, 17, 18, 19, 20, 21, 22, 23] # e-cores (8 cores, 1 thread per core)
}

NUM_RUNS = 30
```

## Número de experimentos

3 aplicações
4 matriz_size
4 CPU sets
30 replicas

3 x 4 x 4 x 30 = 3 x (480) = 1440 

## Criação de arquivo com os experimentos

- Definir valores das variáveis
- Embaralhar 
- Salvar em arquivo .csv

```python
import pandas as pd
import itertools

# --- Definições do Experimento ---
MATRIX_SIZES = [1000, 2000, 4000, 6000, 8000]

CPU_SETS = {
    "8P_HT_OFF": [0, 2, 4, 6, 8, 10, 12, 14],
    "4P_HT_ON": [0, 1, 2, 3, 4, 5, 6, 7],
    "4P_4E_HT_OFF": [0, 2, 4, 6, 16, 17, 18, 19],
    "8E": [16, 17, 18, 19, 20, 21, 22, 23]
}

NUM_RUNS = range(1, 31) # Cria uma lista de 1 a 30 (30 execuções)

# --- Geração das Combinações ---
# itertools.product gera o produto cartesiano de todas as listas.
combinations = itertools.product(MATRIX_SIZES, CPU_SETS.keys(), NUM_RUNS)

# Lista para armazenar os dados de cada experimento
experiment_data = []
experiment_id_counter = 1

for size, cpu_config_name, run_num in combinations:
    # Obtém a lista de núcleos (cores) a partir do nome da configuração
    cpu_cores = CPU_SETS[cpu_config_name]
    
    experiment_data.append({
        "Experiment_ID": f"EXP_{experiment_id_counter:04d}", # EX: EXP_0001
        "Matrix_Size": size,
        "CPU_Set_Name": cpu_config_name,
        "CPU_Cores_List": str(cpu_cores).replace(",", ";"), # Armazena como string, usando ";" como separador opcional
        "Run_Number": run_num
    })
    
    experiment_id_counter += 1

# --- Criação e Exportação do DataFrame ---
df = pd.DataFrame(experiment_data)

# Garante que as colunas sejam ordenadas de forma lógica
df = df[["Experiment_ID", "Matrix_Size", "CPU_Set_Name", "CPU_Cores_List", "Run_Number"]]

output_filename = "experiments_plan.csv"
df.to_csv(output_filename, index=False) # index=False evita escrever a coluna de índice do Pandas no CSV

print(f"✅ Arquivo '{output_filename}' criado com sucesso.")
print(f"Total de experimentos listados: {len(df)}")
```

## Embaralhar

```python
import pandas as pd

# Nome do arquivo CSV criado anteriormente
input_filename = "experiments_plan.csv"

try:
    # 1. Carregar o arquivo CSV
    df = pd.read_csv(input_filename)
    
    # 2. Embaralhar o DataFrame
    # frac=1.0 significa que 100% das linhas serão selecionadas (todas elas).
    # random_state (opcional) garante que o embaralhamento seja o mesmo se você rodar o script novamente, 
    # o que pode ser útil para reprodutibilidade do próprio plano de execução. 
    # Se você quiser um embaralhamento *totalmente* novo a cada rodada, remova 'random_state'.
    df_shuffled = df.sample(frac=1.0, random_state=42).reset_index(drop=True)
    
    # 3. Criar as 3 novas entradas (EDITAR AQUI)
    novas_entradas = [
        {"Experiment_ID": "EXP_WARMUP", "Matrix_Size": 4000, "CPU_Set_Name": "8P_HT_OFF", "CPU_Cores_List": "[0; 2; 4; 6; 8; 10; 12; 14]", "Run_Number": 0}, 
        {"Experiment_ID": "EXP_WARMUP", "Matrix_Size": 4000, "CPU_Set_Name": "8P_HT_OFF", "CPU_Cores_List": "[0; 2; 4; 6; 8; 10; 12; 14]", "Run_Number": 0}, 
        {"Experiment_ID": "EXP_WARMUP", "Matrix_Size": 4000, "CPU_Set_Name": "8P_HT_OFF", "CPU_Cores_List": "[0; 2; 4; 6; 8; 10; 12; 14]", "Run_Number": 0}
    ]
    
    df_novos = pd.DataFrame(novas_entradas)
    
    # 4. Concatenar: Coloca os novos em cima e os embaralhados embaixo
    # ignore_index=True refaz a numeração das linhas de 0 até o fim
    df_final = pd.concat([df_novos, df_shuffled], ignore_index=True)
    
    # 5. Salvar o DataFrame embaralhado no mesmo arquivo
    df_final.to_csv(input_filename, index=False)
    
    print(f"✅ Arquivo '{input_filename}' embaralhado com sucesso!")
    print(f"A nova ordem de execução está pronta.")
    
    print("--- Conteúdo do Arquivo Atualizado ---")
    #pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1200)
    print(df_final)

except FileNotFoundError:
    print(f"❌ Erro: O arquivo '{input_filename}' não foi encontrado. Certifique-se de que o script anterior foi executado.")
except Exception as e:
    print(f"❌ Ocorreu um erro ao processar o arquivo: {e}")
```
