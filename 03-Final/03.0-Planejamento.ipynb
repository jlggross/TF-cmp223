### GIL (Global interpreter Lock)

1. GIL é um *mutex* (mutual exclusion lock) que rpotege o acesso aos objetos internos do Python, impedindo que múltiplas threads executem bytecode Python simultaneamente.

2. O gerenciamento de memória do Python não é *thread-safe*, pois o Python utiliza **Contagem de Referências** (Reference Counting) para gerenciar memória. 
- Caso duas threads tentassem incrementar ou decrementar a contagem de referências de um mesmo objeto ao mesmo tempo, isso poderia causar vazamentos de memória ou corromper dados, levando ao travamento do interpretador.

3. Funcionamento do GIL: Qualquer operação que uma thread precisa exeuctar, seja somar dois números ou acessar um elemento de uma lista, uma thread precisa adquirir o GIL.
- Exemplo: (1) A Thread A adquire o GIL. (2) Executa por um curto período (chamado de quantum ou intervalo de troca). (3) A Thread A libera o GIL. (4) A Thread B adquire o GIL e executa.
- Como resultado prático, a nível do interpretador Python, a execução é **serializada**.
 

###  Multiprocessing vs. Threading

1. Threading
- Todas as threads vivem dentro do mesmo processo do Sistema Operacional e compartilham o mesmo interpretador Python.
- Nível de Paralelismo:
    - **CPU-Bound** (Cálculo pesado): Inexistente (Serializado). Devido ao GIL, apenas uma thread processa Python por vez. 
    - **I/O-Bound** (Rede, Disco): Existe. O GIL é liberado quando a thread faz chamadas de I/O (esperar resposta de DB, ler arquivo). Enquanto uma thread espera, outra processa.
- Compartilhamento de Memória:
    - **Total**. Todas as threads acessam o mesmo espaço de memória (heap).
    - **Vantagem**: Comunicação rápida e fácil entre threads (variáveis globais, objetos passados).
    - **Risco**: Condições de corrida (Race Conditions). É necessário usar **Locks manuais** para evitar que duas threads alterem o mesmo dado de forma inconsistente.

2. Multiprocessing (Módulo multiprocessing)
- Este modelo cria novos processos no Sistema Operacional (fork no Linux/Unix, spawn no Windows/macOS).
- Nível de Paralelismo:
    - Verdadeiro Paralelismo. Cada processo tem sua própria instância do Python e seu próprio GIL. um processo não bloqueia o outro.
- Compartilhamento de Memória:
    - **Isolada** (por padrão). Processos têm espaços de memória separados.
    - **Copy-on-Write** (CoW): Em sistemas Unix (Linux), ao criar um processo filho via fork(), o filho "herda" a memória do pai virtualmente. O SO não duplica a RAM fisicamente **até que o filho tente escrever/modificar algo**.
        - Perfeito para cenários onde os filhos apenas leem da memória

Leitura: O filho lê da mesma memória física que o pai (rápido, baixo uso de RAM).

Escrita: No momento que o filho altera uma variável, o SO copia aquela página de memória para um espaço exclusivo do filho.

Comunicação: Mais lenta e complexa. Requer serialização de dados (pickle) via Pipes, Queues ou memória compartilhada explícita (multiprocessing.Value ou SharedMemory).

### Numpy
- Oversubscription de Threads (NumPy vs. Multiprocessing)
- O Problema: São criados 8 processos e definida a afinidade de cada processo para um núcleo de CPU específico. No entanto, o NumPy (np.dot) por padrão é **multithreaded**. Ele tentará usar várias threads (muitas vezes, todos os núcleos disponíveis) para cada multiplicação de matriz.
- Resultado: Para os 8 processos (do Pool), cada um tentará disparar, digamos, 16 ou 24 threads (do NumPy). Isso é chamado de oversubscription. Os processos e threads competirão caoticamente pelos recursos da CPU, e as medições de tempo serão "poluídas" e inconsistentes.
- A Solução: Forçar o NumPy a usar apenas uma thread dentro de cada processo do pool. Isso garante que o controle de afinidade seja o único responsável pelo paralelismo.
- Como Corrigir: Adicionar as linhas abaixo no início do script (antes de import numpy):

```python
import os
os.environ["OMP_NUM_THREADS"] = "1"
os.environ["MKL_NUM_THREADS"] = "1"
os.environ["OPENBLAS_NUM_THREADS"] = "1"
os.environ["NUMEXPR_NUM_THREADS"] = "1"
```

- Isso diz às bibliotecas de álgebra linear (OpenBLAS, MKL) que o NumPy usa por baixo dos panos para executarem em modo single-thread. Os resultados de benchmark serão drasticamente diferentes, mas agora refletirão corretamente o desempenho do paralelismo gerenciado com multiprocessing.
